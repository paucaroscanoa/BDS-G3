```python
import pandas as pd

%run "D:\\Conocimiento Adquiridos\Data Science _ Codigo Bootcamp\Desarrollo_MP\BDS-G3\Dia_4\pandas_missing_extension.ipynb"

#Probando con nuestro dataset dummy
df = pd.DataFrame.from_dict(
    data = {
        "a": list('asdfasdfas'),
        "b": range(0,10)
    }
)
df.iloc[2:5,0] = None
df.iloc[6:7,1] = None
print('*******************')
print(df)

print(f"total de valores: {df.size}")
print(f"total de valores missing: {df.missing.number_missing()}")
print(f"total de valores completos: {df.missing.number_complete()}")

#Conteo de valores missing
print('*******************')
df.missing.missing_variable_summary()
df.missing.missing_case_summary()
df.missing.missing_variable_table()
df.missing.missing_case_table()
df.missing.missing_variable_plot()
df.missing.missing_case_plot()
```
