```python
import pandas as pd
#creamos un data frame de prueba

df = pd.DataFrame.from_dict(
    data= {
        "a":list('asdfasdfas'),
        "b":range(0,10)
    }
)
print(df)

#Generar valores nulos
df.iloc[2:5,0] = None
df.iloc[6:7,1] = None
print('**************************************')
print(df)

#verificamos valores nulos
print('**************************************')
print(df.isna())

#conteo de nulos
print('**************************************')
print(df.isna().sum())

#conteo total de nuelos
print('**************************************')
print(df.isna().sum().sum())

#conteo por casos
df.isna().pipe(
    lambda df_1: (
        df_1.sum()
        .reset_index(name='n_missing')
        .rename(columns={'index':'variable'})
        .assign(
            n_cases=len(df_1),
            pct_missing=lambda df_2: df_2.n_missing / df_2.n_cases * 100,
        )
    )
)
```
