```python
#IMPUTACIÓN POR PROPAGACIÓN HACIA ADELANTYE Y HACIA ATRAS

import pandas as pd
data = pd.Series([1,None,None,None,5,6])
display(data)

print("------------------PROPAGACIÓN HACIA ADELANTE(FFILL)------------------")
display(data.ffill())

print("------------------PROPAGACIÓN HACIA ATRAS(BFILL)------------------")
display(data.bfill())

print("------------------PROPAGACIÓN POR INTERPOLACIÓN------------------")
display(data.interpolate())
```
