GRAFIOS PERSONALIZADOS

```python
import plotly.express as px
import plotly.graph_objects as go
import pandas as pd
import numpy as np
import seaborn as sns
import plotly.io as pio

df = sns.load_dataset('penguins')
display(df.head(2))
pio.renderers.default = "browser"
```

COLORES Y ESTILOS

```python
fig = px.scatter(df,
                 x='bill_length_mm',
                 y='bill_depth_mm',
                 color='species',
                 title='Bill Length vs Bill Depth',
                 color_discrete_sequence=["red","green","blue"]
)
fig.show()
```

USAR TEMAS DE PLOTLY.IO

```python
import plotly.io as pio
pio.templates
```

PARA CAMBIAR EL TEMA POR DEFECTO

```python
pio.templates.default = 'plotly_dark'
fig = px.scatter(df,
                 x='bill_length_mm',
                 y='bill_depth_mm',
                 color='species',
                 title='Bill Length vs Bill Depth',
                 color_discrete_sequence=["red","yellow","blue"]
)
fig.show()
```

TITULOS , ETIQUETAS Y LEYENDAS

```python

fig = px.scatter(df,
                 x='flipper_length_mm',
                 y='body_mass_g',
                 color='species',
                 color_discrete_sequence=["red","yellow","blue"]
)

fig.update_layout(
    title='Masa Corporal vs Logintud de Aleta',
    xaxis_title ='Logintud de Aleta(mm)',
    yaxis_title ='Masa Corporal(g)',
    legend_title='Especie de Pingüino'
)

fig.show()
```

USO DE ESCALAS , COLORES Y TEMAS

### Opciones de color para escalas: "Viridis", "Cividis", "Plasma", "Blues", "Greens"

```python

fig = px.density_heatmap(df,
                        x='bill_length_mm',
                        y='bill_depth_mm',
                        title='Bill Length vs Bill Depth',
                        color_continuous_scale="Viridis")
fig.show()
```

PERSONALIZANDO COLORES PARA VALORES DISCRETOS

```python

fig = px.scatter(df,
                x='bill_length_mm',
                y='bill_depth_mm',
                title='Bill Length vs Bill Depth',
                color='species',
                color_discrete_map={"Adelie":"purple","Gentoo":"orange","Chinstrap":"cyan"})
fig.show()
     
```

ANOTACIONES Y TEXTOS EN GRAFICOS

#### COMO RESALTAR UN PUNTO EN PARTICULAR CON ANOTACIONES

```python

fig = px.scatter(df,
                x='flipper_length_mm',
                y='body_mass_g',
                title='Masa Corporal vs Logintud de Aleta',
                color='species',
                color_discrete_map={"Adelie":"purple","Gentoo":"orange","Chinstrap":"cyan"})

fig.add_annotation(
    x=210,y=4500,
    text='Ejemplo : Pinguino con aleta de 210',
    showarrow=True,
    arrowhead=2
)

fig.show()
```

ETIQUETAR MÚLTIPLES PUNTOS EN UN GRÁFICO

```python
fig = px.scatter(df,
                x='bill_length_mm',
                y='bill_depth_mm',
                title='Bill Length vs Bill Depth',
                color='species',
                color_discrete_map={"Adelie":"purple","Gentoo":"orange","Chinstrap":"cyan"})

for i in range(0,len(df),20):
  fig.add_annotation(
      x=df['bill_length_mm'][i],
      y=df['bill_depth_mm'][i],
      text=df['species'][i],
      showarrow=False,
      font=dict(size=10,color="black"),
      bgcolor='lightyellow'
  )

fig.show()
```
