GRAFICOS AVANZADOS

```python
import pandas as pd
import seaborn as sns
import plotly.io as pio
import plotly.express as px

df = sns.load_dataset('penguins')
df.head(2)
pio.renderers.default = "browser"
```

GRÁFICO DE AREAS

```python
df_sorted = df.sort_values(by='body_mass_g')

fig = px.area(df_sorted,
              x='body_mass_g',
              y='flipper_length_mm',
              color='species',
              title='Longitud de Aleta vs Masa Corporal(Area de Especie)')
fig.show()
```

GRÁFICO DE DENSIDAD

```python
fig = px.density_contour(df,
                         x='body_mass_g',
                         y='flipper_length_mm',
                         color='species',
                         title='Densidad de Masa Corporal vs Longitud de Aleta')
fig.show()
```

BOX PLOT

```python

fig = px.box(df,
             x='species',
             y='body_mass_g',
             color='species',
             title= 'Distribución de Peso por Especie')
fig.show()
     
```

GRÁFICO DE VIOLIN

```python
fig = px.violin(df,
                x='species',
                y='body_mass_g',
                color='species',
                box=True,
                points='all',
                title='Ditribución del Peso por Especie')
fig.show()
```

MAPA DE CALOR

```python
import plotly.figure_factory as ff
import numpy as np

correlation = df.corr(numeric_only=True)
correlation

fig = ff.create_annotated_heatmap(
    z=correlation.values,
    x=list(correlation.columns),
    y=list(correlation.index),
    colorscale='Viridis',
    annotation_text=np.round(correlation.values,2)
)
fig.show()
```

GRÁFICOS 3D

```python
fig = px.scatter_3d(df,
                    x='bill_length_mm',
                    y='bill_depth_mm',
                    z='body_mass_g',
                    color='species',
                    title='Relación 3D entre bill_length bill_depth y body_mass')
fig.show()
```
