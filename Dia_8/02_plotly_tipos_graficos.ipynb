TIPO DE GRAFICOS

```python
import pandas as pd
import seaborn as sns
import plotly.io as pio
import plotly.express as px

df = sns.load_dataset('penguins')
df.head(2)
pio.renderers.default = "browser"
```

GRÁFICO DE DISPERSIÓN (SCATTER)

```python
fig = px.scatter(df,
                 x='flipper_length_mm',
                 y='body_mass_g',
                 color='species',
                 title='Relación entre flipper_length y body_mass')
fig.show()
```

GRÁFICO DE BARRAS( BAR)

```python
fig = px.bar(df,
             x='species',
             title='Cantidad x Especie de Pinguino')
fig.show()
```

HISTOGRAMA (histogram)

```python
fig = px.histogram(df,
                   x='body_mass_g',
                   color='species',
                   title='Distribución de Peso por especie de pingüino')
fig.show()
```

GRÁFICO DE LINEAS(LINE)

```python
df_sorted = df.sort_values(by=['island','body_mass_g'])
fig = px.line(df_sorted,
              x='island',
              y='body_mass_g',
              color='species',
              title='variación de peso por Isla')
fig.show()
```

GRÁFICO DE PIE ( PIE)

```python
fig = px.pie(df,
             names='species',
             title='Proporción de pinguinos por Especie')
fig.show()
```
