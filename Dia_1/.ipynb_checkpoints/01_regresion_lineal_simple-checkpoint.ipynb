### USO DE SCIPY

Para probar tu regresión lineal con valores que no están en tu conjunto de datos (es decir, para hacer predicciones), puedes utilizar la ecuación de la línea de regresión que has obtenido de scipy.stats.linregress. La ecuación es de la forma:

𝑦 = 𝑚 𝑥 +𝑏

donde:

1. m es la pendiente (slope)
2. b es el intercepto (intercept)
3. x es el valor nuevo para el cual quieres predecir

```python
import scipy.stats

# datos de ejemplo
x = np.array([0,1,2,3,4])
y = np.array([2,2.8,3.6,4.4,5.2])
print(x,y)
```

### GRAFICAMOS LA RELACIÓN ENTRE X y Y

```python
plt.figure(figsize=(10,6))
plt.scatter(x,y,color='blue',label='Datos Observados')
```

### REALIZAMOS LA REGRESIÓN LINEAL SIMPLE

```python
result = scipy.stats.linregress(x,y)
result
```

### MOSTRAMOS RESULTADOS DE LA REGRESIÓN LINEAL

```python
print(f'Pendiente(slope): {result.slope}')
print(f'Intercepto(intercept): {result.intercept}')
print(f'Coeficiente de Correlación(rvalue) {result.rvalue}')
print(f'Valor p(pvalue) : {result.pvalue}')
print(f'Error stándar de la pendiente (stderr) : {result.stderr}')
```

```python
y_pred = result.slope * x + result.intercept
y_pred
```

### GRAFICAMOS LA REGRESIÓN LINEAL

```python
plt.figure(figsize=(10,6))
plt.scatter(x,y,color='blue',label='Datos Observados')
plt.plot(x,y_pred,color='red',label='Línea de regresión')

plt.xlabel('X')
plt.ylabel('Y')
plt.title('Regresión lineal con scipy')
plt.legend()
plt.show()
```

### PROBAMOS NUESTRA PREDICCIÓN

```python
x_new = 6
y_new_pred = result.slope * x_new + result.intercept
print(f'predicción para x = {x_new}: y = {y_new_pred}')
```

### GRAFICAMOS EL VALOR DE PREDICCIÓN

```python

m = result.slope
b = result.intercept
y_pred_x = m * x + b
plt.figure(figsize=(10,6))
plt.scatter(x,y,color='blue',label='datos observador')
plt.xlabel('x')
plt.ylabel('y')
plt.plot(x,y_pred_x,color='red',label='Linea de regresión')
plt.scatter(x_new,y_new_pred,color='green',marker='o',s=100,label=f'predicción para x={x_new}')

plt.legend()
plt.show()
```

```python
import numpy as np
import matplotlib.pyplot as plt
from scipy import stats

# Datos de ejemplo
x = np.array([1, 2, 3, 4, 5])
y = np.array([2, 4, 5, 4, 5])

#graficamos la relación entre x e y

plt.figure(figsize=(10,6))
plt.scatter(x,y,color='blue',label='Datos Observados')

# Ajustar la regresión lineal con scipy.stats.linregress
regresion = stats.linregress(x, y)
slope = regresion.slope
intercept = regresion.intercept
Coeficiente= result.rvalue
Valor_p=result.pvalue
Error_standar=result.stderr

print("**********************************")
print("Resultados:")
print("Pendiente:", slope)
print("Intercepto:", intercept)
print("Coeficiente Correlación:", Coeficiente)
print("P_Value):", Valor_p)
print("Error Estandar):", Error_standar)

# Número de observaciones y cálculo del error estándar residual
n = len(x)
residuos = y - (intercept + slope * x)
s_err = np.sqrt(np.sum(residuos**2) / (n - 2))

# Media de x y suma de cuadrados
x_mean = np.mean(x)
Sxx = np.sum((x - x_mean)**2)

# --- Predicción para un nuevo valor fuera del rango ---
# Definimos un nuevo valor (por ejemplo, x = 6, siendo el rango original de 1 a 5)
new_x = 6
new_y = intercept + slope * new_x  # Predicción para el nuevo valor
print('***********************************')
print(f'predicción para x = {new_x}: y = {new_y}')

# Calcular el error estándar para la predicción de una nueva observación en new_x
se_pred_new = s_err * np.sqrt(1 + 1/n + (new_x - x_mean)**2 / Sxx)

# Valor crítico t para un 95% de confianza (alpha=0.05) con n-2 grados de libertad
alpha = 0.05
t_val = stats.t.ppf(1 - alpha/2, n - 2)

# Límites del intervalo de predicción para el nuevo valor
pred_lower_new = new_y - t_val * se_pred_new
pred_upper_new = new_y + t_val * se_pred_new

# --- Para graficar la línea de regresión y el intervalo de predicción ---
# Generamos puntos entre el mínimo de x y el nuevo valor para visualizar la extrapolación
x_pred = np.linspace(x.min(), new_x, 100)
y_pred = intercept + slope * x_pred

# Calcular el error estándar de la predicción para cada punto (incluye error en la estimación y variabilidad individual)
se_pred = s_err * np.sqrt(1 + 1/n + (x_pred - x_mean)**2 / Sxx)
pred_lower = y_pred - t_val * se_pred
pred_upper = y_pred + t_val * se_pred

# --- Gráfica ---
plt.figure(figsize=(10, 6))
plt.scatter(x, y, color='blue', label='Datos observados')
plt.plot(x_pred, y_pred, color='red', label='Línea de regresión')
plt.fill_between(x_pred, pred_lower, pred_upper, color='green', alpha=0.3,
                 label='Intervalo de predicción (95%)')

# Graficar la predicción para el nuevo valor fuera del rango
plt.scatter(new_x, new_y, color='orange', s=100, label='Nueva predicción')
plt.errorbar(new_x, new_y, yerr=t_val * se_pred_new, fmt='o', color='orange', capsize=10,
             label='Intervalo para nueva predicción')

plt.xlabel('x')
plt.ylabel('y')
plt.title('Predicción y intervalo de predicción para un valor fuera del rango')
plt.legend()
plt.show()
```
