```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
import nhanes.load
import sklearn.compose
import sklearn.preprocessing
import janitor

#Cargamos nuestro dataset
display(nhanes.load.load_NHANES_data(year="2017-2018"))

nhanes_raw_df = (
    nhanes.load.load_NHANES_data(year="2017-2018")
    .clean_names(
        case_type="snake"
    )
)
display(nhanes_raw_df.head(2))

#Creamos un datagrame más pequeño
nhanes_df =  (
    nhanes_raw_df.select_columns(
        "general_health_condition",
        "age_in_years_at_screening",
        "gender"
    )
).rename_columns(
    {
        "age_in_years_at_screening":"age"
    }
)
display(nhanes_df)

#Codificación ordinal con sklearn 
X = [["Male"],["Female"],["Female"]]
display(X)

encoder = sklearn.preprocessing.OrdinalEncoder()
encoder.fit_transform(X)
encoder.categories_
encoder.inverse_transform([[1],[0],[0]])

nhanes_df.dtypes

categorical_columns=nhanes_df.select_dtypes(object).columns
display(categorical_columns)

#Creamos un transformador para la codificación

categorical_transformer = sklearn.compose.make_column_transformer(
    (sklearn.preprocessing.OrdinalEncoder(),categorical_columns),
    remainder="passthrough"
)

#Aplicamos el transformador a nuestro dataset
nhanes_transformed_df = (
    pd.DataFrame(
        categorical_transformer.fit_transform(nhanes_df),
        columns = categorical_transformer.get_feature_names_out(),
        index=nhanes_df.index
    )
    .rename_columns(
        function = lambda x:x.removeprefix("ordinalencoder__")
    )
    .rename_columns(
        function = lambda x:x.removeprefix("remainder__")
    )
)
display(nhanes_transformed_df)
```
